```python
import matplotlib.pyplot as plt
import numpy as np

# Data
languages = ["English", "Hindi", "Marathi"]
deletion_auc = [0.1571, 0.2386, 0.2581]
corr_vals = [-0.0082, -0.0011, 0.1811]

x = np.arange(len(languages))
width = 0.35  # bar width

plt.figure(figsize=(7,5))

bars1 = plt.bar(x - width/2, deletion_auc, width, label="Deletion AUC", color="#4c72b0", alpha=0.85)
bars2 = plt.bar(x + width/2, corr_vals, width, label="Attn–Grad Corr", color="#dd8452", alpha=0.85)

# Annotate values
for bar, val in zip(bars1, deletion_auc):
    plt.text(bar.get_x() + bar.get_width()/2, val + 0.01,
             f"{val:.3f}", ha='center', va='bottom', fontsize=11)
for bar, val in zip(bars2, corr_vals):
    offset = 0.02 if val >= 0 else -0.02
    va = 'bottom' if val >= 0 else 'top'
    plt.text(bar.get_x() + bar.get_width()/2, val + offset,
             f"{val:.3f}", ha='center', va=va, fontsize=11)

plt.axhline(0, color="gray", linewidth=0.8, linestyle="--")
plt.xticks(x, languages, fontsize=12)
plt.yticks(fontsize=12)
plt.ylabel("Metric Value", fontsize=13)
plt.title("Explainability Metrics Across Languages", fontsize=14, pad=12)
plt.legend(fontsize=11)
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Set limits to include negative values clearly
plt.ylim(-0.1, 0.3)

plt.tight_layout()
plt.savefig("explainability_combined_bar.png", dpi=400, bbox_inches="tight")
plt.show()
```
